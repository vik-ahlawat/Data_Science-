```python
# Linear algebra 

import numpy as np

# Define the data vectors
X1 = np.array([[8], [4], [7]])
X2 = np.array([[2], [8], [1]])
X3 = np.array([[3], [1], [1]])
X4 = np.array([[9], [7], [4]])

# Store them in a list for easier iteration
data_vectors = [X1, X2, X3, X4]
n = len(data_vectors)

# --- Step 1: Calculate the empirical or sample mean (X_bar) ---
# Sum all X^(i) vectors
sum_X = np.zeros_like(X1)  # Initialize with zeros of the same shape as X1
for X_i in data_vectors:
    sum_X += X_i

# Calculate the mean
X_bar = (1 / n) * sum_X

print("Sample Mean (X_bar):")
print(X_bar)

# --- Step 2: Calculate the empirical covariance matrix (S) ---
# Calculate the sum of X^(i) * (X^(i))^T
sum_outer_products = np.zeros((X1.shape[0], X1.shape[0])) # Initialize a 3x3 matrix of zeros
for X_i in data_vectors:
    sum_outer_products += np.outer(X_i, X_i) # outer product of X_i and X_i

# Calculate X_bar * (X_bar)^T
outer_product_X_bar = np.outer(X_bar, X_bar)

# Calculate S
S = (1 / n) * (sum_outer_products - n * outer_product_X_bar)

print("\nEmpirical Covariance Matrix (S):")
print(S)
```

```python
# Define the unit vector u
u = (1/np.sqrt(5)) * np.array([1, 2])

# Define the data vectors
X1 = np.array([1, 2])
X2 = np.array([3, 4])
X3 = np.array([-1, 0])

# Calculate the dot products u_dot_Xi
u_dot_X1 = np.dot(u, X1)
u_dot_X2 = np.dot(u, X2)
u_dot_X3 = np.dot(u, X3)

# Store the results in a list
data_set = [u_dot_X1, u_dot_X2, u_dot_X3]

# Calculate the mean of the data set
mean_data_set = np.mean(data_set)

# Calculate the empirical variance
# Using ddof=0 for empirical variance (dividing by N)
empirical_variance = np.var(data_set, ddof=0) 

print(f"Mean of the data set: {mean_data_set}")
print(f"Empirical variance of the data set: {empirical_variance}")
```
