```
 # --- 2. CENTRALITY CALCULATION & AGGREGATION ---
all_centrality_data = []

# Iterate through the phases that were successfully loaded into G
for phase_num, graph in G.items():
    # Initialize dictionary for centralities in this phase
    centrality_measures = {}

    # Calculate Degree and Betweenness Centrality (normalized by default)
    centrality_measures['Degree'] = nx.degree_centrality(graph)
    centrality_measures['Betweenness'] = nx.betweenness_centrality(graph)

    # Calculate Eigenvector Centrality with error handling
    try:
        centrality_measures['Eigenvector'] = nx.eigenvector_centrality(graph)
    except Exception:
        # Assign 0.0 to all nodes if computation fails (e.g., disconnected components)
        centrality_measures['Eigenvector'] = {node: 0.0 for node in graph.nodes()}

    # Store results in the master list
    for measure, results in centrality_measures.items():
        for player, value in results.items():
            all_centrality_data.append({
                'Player': player,
                'Phase': phase_num,
                'Centrality Type': measure,
                'Centrality Value': value
            })

# Create the master DataFrame
df_centrality = pd.DataFrame(all_centrality_data)
 # COMPUTE MEAN CENTRALITY AND FIND TOP 3 ---
output_dfs = []

if not df_centrality.empty:
    # Group by Centrality Type and Player, then calculate the mean across all phases
    df_mean_centrality = df_centrality.iloc[:, :5].groupby(['Centrality Type', 'Player'])['Centrality Value'].mean().reset_index()

    # Find the top 3 players for each centrality type
    for centrality_type in df_mean_centrality['Centrality Type'].unique():
        df_subset = df_mean_centrality[df_mean_centrality['Centrality Type'] == centrality_type]

        # Sort and get top 3 players
        top_6 = df_subset.sort_values(by='Centrality Value', ascending=False).head(6)

        # Format the result for printing
        top_players = top_6[['Player', 'Centrality Value']].rename(
            columns={'Centrality Value': f'Mean {centrality_type} Centrality'}
        )
        top_players['Rank'] = [1, 2, 3,4,5,6]

        output_dfs.append(top_players[['Rank', 'Player', f'Mean {centrality_type} Centrality']])

    # Combine the top results into a single formatted output string
    output_string = "## Top 6 Players with Highest Centrality Mean Across 11 Phases\n"
    for df in output_dfs:
        centrality_type = df.columns[-1].replace('Mean ', '').replace(' Centrality', '')
        output_string += f"\n### Mean {centrality_type} Centrality\n"
        output_string += df.to_markdown(index=False, floatfmt=".4f")
        output_string += "\n"

    print(output_string)

else:
    print("Error: No network data was successfully loaded to perform centrality calculations.")
```

```


def rank_centralities(graph):
    """
    Calculates and ranks nodes by betweenness and eigenvector centrality.

    Args:
        graph (nx.Graph or nx.DiGraph): The graph to analyze.

    Returns:
        tuple: A tuple containing two lists of tuples:
               - betweenness_ranked (list): Nodes ranked by betweenness centrality
                                            in descending order.
                                            Each tuple is (node, centrality_score).
               - eigenvector_ranked (list): Nodes ranked by eigenvector centrality
                                            in descending order.
                                            Each tuple is (node, centrality_score).
    """

    # Calculate betweenness centrality
    betweenness_centrality = nx.betweenness_centrality(graph)

    # Calculate eigenvector centrality
    eigenvector_centrality = nx.eigenvector_centrality(graph)

    # Rank nodes by betweenness centrality (descending order)
    betweenness_ranked = sorted(betweenness_centrality.items(),
                                key=lambda item: item[1],
                                reverse=True)

    # Rank nodes by eigenvector centrality (descending order)
    eigenvector_ranked = sorted(eigenvector_centrality.items(),
                                key=lambda item: item[1],
                                reverse=True)

    return betweenness_ranked, eigenvector_ranked

if __name__ == '__main__':

    # Calculate and rank centralities
    betweenness_rank, eigenvector_rank = rank_centralities(G[9])

    print("Nodes ranked by Betweenness Centrality:")
    for node, score in betweenness_rank:
        print(f"Node {node}: {score:.4f}")

    print("\nNodes ranked by Eigenvector Centrality:")
    for node, score in eigenvector_rank:
        print(f"Node {node}: {score:.4f}")
```

```
import pandas as pd
import networkx as nx
import sys

# --- 1. SETUP AND DATA LOADING (Directed Graphs) ---
G = {}
base_url = "https://raw.githubusercontent.com/ragini30/Networks-Homework/main/"

# Loop through phases 1 to 11 to load data as Directed Graphs (DiGraph)
for i in range(1, 12):
  var_name = "phase" + str(i)
  file_name = base_url + var_name + ".csv"

  try:
    df = pd.read_csv(file_name, index_col=["players"])
  except Exception as e:
    print(f"Warning: Failed to read {file_name}. Skipping phase {i}.", file=sys.stderr)
    continue

  df.columns = "n" + df.columns
  df.index = df.columns

  df_binary = df.copy()
  df_binary[df_binary > 0] = 1

  # Create the NetworkX Directed Graph (DiGraph)
  G[i] = nx.from_pandas_adjacency(df_binary, create_using=nx.DiGraph())

# --- 2. HITS CALCULATION & TRACKING ---
# We will track the scores for the top 3 critical players identified previously: Node 1, Node 3, and Node 12.
# Node 85 is less active but highly influential; we will track it as well.
tracked_players = ['n1', 'n3', 'n12', 'n85']
tracking_data = []

# Incident points: P4 (Seizure 1), P6 (Seizure 3), P9 (Seizure 2)
incident_phases = [4, 6, 9]

# Iterate through the phases
for phase_num, graph in G.items():

    # Compute HITS scores
    try:
        # max_iter=1000000 as requested for consistency
        hubs, authorities = nx.hits(graph, max_iter=1000000)
    except Exception:
        hubs = {node: 0.0 for node in graph.nodes()}
        authorities = {node: 0.0 for node in graph.nodes()}


    # Store data for tracked players
    for player in tracked_players:
        if player in graph.nodes():
            is_incident = phase_num in incident_phases

            tracking_data.append({
                'Phase': phase_num,
                'Player': player,
                'Hub Score': hubs.get(player, 0.0),
                'Authority Score': authorities.get(player, 0.0),
                'Is_Incident_Phase': is_incident
            })

df_tracking = pd.DataFrame(tracking_data)

# --- 3. RESULTS AND VISUALIZATION ---

# Find the overall top Authority and Hub for reporting context
df_all_hubs = df_tracking[df_tracking['Player'] != 'n85'].groupby('Player')['Hub Score'].mean().sort_values(ascending=False).head(1)
df_all_authorities = df_tracking.groupby('Player')['Authority Score'].mean().sort_values(ascending=False).head(1)

# Prepare data for plotting the evolution of key players
df_plot = df_tracking.pivot(index='Phase', columns='Player', values='Authority Score')
df_plot_hubs = df_tracking.pivot(index='Phase', columns='Player', values='Hub Score')

# Use Matplotlib for generating the required figure
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid')

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Define the incident markers for the plot
incident_phases_x = [p for p in incident_phases if p in df_tracking['Phase'].unique()]

# Plot 1: Authority Score Evolution
for col in df_plot.columns:
    axes[0].plot(df_plot.index, df_plot[col], marker='o', linestyle='-', label=col)

# Add vertical lines for incident phases
for x in incident_phases_x:
    axes[0].axvline(x=x, color='gray', linestyle='--', alpha=0.7)
    axes[0].text(x, axes[0].get_ylim()[1]*0.95, f'P{x} Seizure', rotation=90, color='red', fontsize=9, ha='right')

axes[0].set_title('Evolution of Authority Centrality (Ultimate Command)')
axes[0].set_xlabel('Phase')
axes[0].set_ylabel('Authority Score')
axes[0].set_xticks(df_plot.index)
axes[0].legend(title='Player')

# Plot 2: Hub Score Evolution
for col in df_plot_hubs.columns:
    axes[1].plot(df_plot_hubs.index, df_plot_hubs[col], marker='s', linestyle='-', label=col)

# Add vertical lines for incident phases
for x in incident_phases_x:
    axes[1].axvline(x=x, color='gray', linestyle='--', alpha=0.7)
    axes[1].text(x, axes[1].get_ylim()[1]*0.95, f'P{x} Seizure', rotation=90, color='red', fontsize=9, ha='right')

axes[1].set_title('Evolution of Hub Centrality (Operational Direction)')
axes[1].set_xlabel('Phase')
axes[1].set_ylabel('Hub Score')
axes[1].set_xticks(df_plot_hubs.index)
axes[1].legend(title='Player')

plt.tight_layout()
plt.savefig('authority_hub_evolution_plot.png')

# Output Key Findings Table (Authority Scores at Critical Junctures)
critical_scores = df_tracking[df_tracking['Phase'].isin([3, 4, 5, 8, 9, 10])]
critical_scores_pivot = critical_scores.pivot(index='Player', columns='Phase', values='Authority Score').fillna(0).sort_index()

print("## Critical Authority Score Changes Around Seizures\n")
print("Scores show the Authority of key players immediately before (P3, P8) and after (P5, P10) major seizures (P4, P9).\n")
print(critical_scores_pivot.to_markdown(floatfmt=".4f"))

# Final result placeholder for the report content
print(f"\nOverall Top Authority: {df_all_authorities.index[0]} (Mean Score: {df_all_authorities.iloc[0]:.4f})")
print(f"Overall Top Hub: {df_all_hubs.index[0]} (Mean Score: {df_all_hubs.iloc[0]:.4f})")
```

```
from networkx.algorithms import community
def compute_stats(G):
    stats = {}
    stats['nodes'] = len(G)
    stats['edges'] = G.number_of_edges()
    stats['density'] = nx.density(G)
    stats['clustering'] = nx.average_clustering(G)

    # Work on largest connected component for distance measures
    if nx.is_connected(G):
        subG = G
    else:
        subG = G.subgraph(max(nx.connected_components(G), key=len))

    stats['avg_path_length'] = nx.average_shortest_path_length(subG)
    stats['diameter'] = nx.diameter(subG)

    # Degree and betweenness centralization
    deg_centrality = nx.degree_centrality(G)
    bet_centrality = nx.betweenness_centrality(G)
    stats['degree_centralization'] = max(deg_centrality.values())
    stats['betweenness_centralization'] = max(bet_centrality.values())

    # Community detection (greedy modularity)
    communities = list(community.greedy_modularity_communities(G))
    stats['num_communities'] = len(communities)
    stats['modularity'] = community.modularity(G, communities)

    return stats

# Compute for all 11 phases
results = {f"Phase {i}": compute_stats(G[i]) for i in range(1, 12)}
df = pd.DataFrame(results).T
df
```
